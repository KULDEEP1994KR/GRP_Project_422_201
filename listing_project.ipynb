```python
df_l = pd.read_csv("listings_last_week.csv")
```

```python
df_l.sample(5)
```

```python
import matplotlib.pyplot as plt
# plt.figure(figsize=(12, 7))
plt.hist(df_l["price"].dropna(), bins=60, range=(0, 1000),edgecolor='black')
# plt.xlim(0, 8000)
# plt.ylim(0, 150000) 


plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Price - New Zealand")
plt.show()
```

```python
chrch_df = df_l[df_l["neighbourhood_group"] == "Christchurch City"]

plt.hist(chrch_df["price"].dropna(), bins=50, range=(0, 1000), edgecolor='black')
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title(" Price - Christchurch City")
plt.show()
```

```python
chrch_df.describe
```

```python
chrch_df.dtypes
```

```python
chrch_df["last_review"] = pd.to_datetime(chrch_df["last_review"])
```

```python
chrch_df.dtypes
```

```python
chrch_df["last_review"].isna().sum()
```

```python
# Convert review column to datetime
chrch_df["last_review"] = pd.to_datetime(chrch_df["last_review"])

# Create publish date column
chrch_df["publish_date"] = pd.to_datetime("16-06-2026", dayfirst=True)

# Calculate difference
chrch_df["date_difference"] = (chrch_df["publish_date"] - chrch_df["last_review"]).dt.days.astype("Int64")  # Use Int64 to allow for NaN values
```

```python
chrch_df.head(3)
```

```python
chrch_df.dtypes
```

```python
plt.hist(chrch_df["date_difference"].dropna(), bins=50, range=(0, 1000), edgecolor='black')
plt.xlabel("Date Difference")
plt.ylabel("Frequency")
plt.title("Date Difference - Christchurch City")
plt.show()
```

```python
df_sorted = chrch_df.sort_values("date_difference", ascending=False)
top_10 = df_sorted.head(int(len(df_sorted) * 0.10))
```
