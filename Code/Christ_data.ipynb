```python
Q1 = df_all_christ['price'].quantile(0.25)
Q3 = df_all_christ['price'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
outliers = df_all_christ[
    (df_all_christ['price'] < lower_bound) |
    (df_all_christ['price'] > upper_bound)
]

outliers[['name', 'room_type', 'neighbourhood', 'price']].head()
```

```python
price_before = df_all_christ['price'].copy()
```

```python
df_all_christ['price'] = (
    df_all_christ
    .groupby(['room_type', 'neighbourhood'])['price']
    .transform(lambda x: x.fillna(x.median()))
)
```

```python
 df_all_christ['price'].isna().sum()
```

```python
print(price_before.describe())


print(df_all_christ['price'].describe())
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
     [price_before.dropna(), df_all_christ['price'].dropna()],
     tick_labels=['Before', 'After']
)

plt.ylabel('Price')
plt.title('Price Distribution Before and After Imputation')
plt.show()
```

```python
df_all_christ[['reviews_per_month', 'last_review','Difference_date']] = (
    df_all_christ[['reviews_per_month', 'last_review','Difference_date']].fillna(0)
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.hist(
    price_before.dropna(),
    bins=50,
    alpha=0.5,
    label='Before imputation',
    range=(0, 1000)
)

plt.hist(
    df_all_christ['price'],
    bins=50,
    alpha=0.5,
    label='After imputation',
    range=(0, 1000)
)

plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Price Distribution Before and After Imputation')
plt.legend()

plt.show()
```
