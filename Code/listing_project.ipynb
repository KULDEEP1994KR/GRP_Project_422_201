```python
import pandas as pd
```

```python
oct_data = pd.read_csv("../Data/listings_Oct_25.csv")
nov_data = pd.read_csv("../Data/listings_Nov_25.csv")
dec_data = pd.read_csv("../Data/listings_Dec_25.csv")
jan_data = pd.read_csv("../Data/listings_Jan_26.csv")
feb_data = pd.read_csv("../Data/listings_Feb_26.csv")
mar_data = pd.read_csv("../Data/listings_Mar_26.csv")
apr_data = pd.read_csv("../Data/listings_Apr_26.csv")
may_data = pd.read_csv("../Data/listings_may_26.csv")
jun_data = pd.read_csv("../Data/listings_Jun_26.csv")
listing_july= pd.read_csv("../Data/listings_july.csv")
listing_aug= pd.read_csv("../Data/listings_aug.csv")
```

```python
oct_data["Pub_date"] = pd.to_datetime("2025-10-07").date()
nov_data["Pub_date"] = pd.to_datetime("2025-11-07").date()   
dec_data["Pub_date"] = pd.to_datetime("2025-12-11").date()
jan_data["Pub_date"] = pd.to_datetime("2026-01-16").date()
feb_data["Pub_date"] = pd.to_datetime("2026-02-13").date()
mar_data["Pub_date"] = pd.to_datetime("2026-03-17").date()
apr_data["Pub_date"] = pd.to_datetime("2026-04-16").date()    
may_data["Pub_date"] = pd.to_datetime("2026-05-23").date()
jun_data["Pub_date"] = pd.to_datetime("2026-06-19").date()
listing_july["Pub_date"] = pd.to_datetime("2026-07-12").date()
listing_aug["Pub_date"] = pd.to_datetime("2026-08-13").date()
```

```python
all_listings = pd.concat(
    [oct_data, nov_data, dec_data, jan_data, feb_data,
     mar_data, apr_data, may_data, jun_data, listing_july, listing_aug],
    ignore_index=True
)
```
