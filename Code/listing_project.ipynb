```python
import pandas as pd
```

```python
oct_data = pd.read_csv("../Data/listings_Oct_25.csv")
nov_data = pd.read_csv("../Data/listings_Nov_25.csv")
dec_data = pd.read_csv("../Data/listings_Dec_25.csv")
jan_data = pd.read_csv("../Data/listings_Jan_26.csv")
feb_data = pd.read_csv("../Data/listings_Feb_26.csv")
mar_data = pd.read_csv("../Data/listings_Mar_26.csv")
apr_data = pd.read_csv("../Data/listings_Apr_26.csv")
may_data = pd.read_csv("../Data/listings_may_26.csv")
jun_data = pd.read_csv("../Data/listings_Jun_26.csv")
listing_july= pd.read_csv("../Data/listings_july.csv")
listing_aug= pd.read_csv("../Data/listings_aug.csv")
```

```python
oct_data["Pub_date"] = pd.to_datetime("2025-10-07").date()
nov_data["Pub_date"] = pd.to_datetime("2025-11-07").date()   
dec_data["Pub_date"] = pd.to_datetime("2025-12-11").date()
jan_data["Pub_date"] = pd.to_datetime("2026-01-16").date()
feb_data["Pub_date"] = pd.to_datetime("2026-02-13").date()
mar_data["Pub_date"] = pd.to_datetime("2026-03-17").date()
apr_data["Pub_date"] = pd.to_datetime("2026-04-16").date()    
may_data["Pub_date"] = pd.to_datetime("2026-05-23").date()
jun_data["Pub_date"] = pd.to_datetime("2026-06-19").date()
listing_july["Pub_date"] = pd.to_datetime("2026-07-12").date()
listing_aug["Pub_date"] = pd.to_datetime("2026-08-13").date()
```

```python
oct_data.head()
```

```python
all_listings = pd.concat(
    [oct_data, nov_data, dec_data, jan_data, feb_data,
     mar_data, apr_data, may_data, jun_data, listing_july, listing_aug],
    ignore_index=True
)
```

```python
all_listings.head()
```

```python
oct_data_chrch = oct_data[oct_data["neighbourhood_group"] == "Christchurch City"].copy()
nov_data_chrch = nov_data[nov_data["neighbourhood_group"] == "Christchurch City"].copy()
dec_data_chrch = dec_data[dec_data["neighbourhood_group"] == "Christchurch City"].copy()
jan_data_chrch = jan_data[jan_data["neighbourhood_group"] == "Christchurch City"].copy()
feb_data_chrch = feb_data[feb_data["neighbourhood_group"] == "Christchurch City"].copy()
mar_data_chrch = mar_data[mar_data["neighbourhood_group"] == "Christchurch City"].copy()
apr_data_chrch = apr_data[apr_data["neighbourhood_group"] == "Christchurch City"].copy()
may_data_chrch = may_data[may_data["neighbourhood_group"] == "Christchurch City"].copy()
jun_data_chrch = jun_data[jun_data["neighbourhood_group"] == "Christchurch City"].copy()
july_data_chrch = listing_july[listing_july["neighbourhood_group"] == "Christchurch City"].copy()
aug_data_chrch = listing_aug[listing_aug["neighbourhood_group"] == "Christchurch City"].copy()
```

```python
oct_data_chrch.head()
```

```python
jun_data_chrch.head()
```

```python
all_chrch = pd.concat(
    [
        oct_data_chrch,
        nov_data_chrch,
        dec_data_chrch,
        jan_data_chrch,
        feb_data_chrch,
        mar_data_chrch,
        apr_data_chrch,
        may_data_chrch,
        jun_data_chrch,
        july_data_chrch,
        aug_data_chrch
    ],
    ignore_index=True
)
```

```python
all_chrch["last_review"] = pd.to_datetime(all_chrch["last_review"])
all_chrch["Pub_date"] = pd.to_datetime(all_chrch["Pub_date"])
```

```python
all_chrch.dtypes
```

```python
all_chrch["Difference_date"] = (all_chrch["Pub_date"] - all_chrch["last_review"]).dt.days.astype("Int64")  # Use Int64 to allow for NaN values
```
