## Converting bond data set and Air Bnb data set area_code coloumn type into string so can match the dtype of both coloumns.

```python
df_tnc["area_code"] = pd.to_numeric(
    df_tnc["area_code"], errors="coerce"
).astype("Int64")
```

```python
christ_area["area_code"] = pd.to_numeric(
    christ_area["area_code"], errors="coerce"
).astype("Int64")

```

## Preparing time fields
Dates are converted into a common quarterly format:
- `Pub_date` is converted to a datetime
- `TimeFrame` is created/standardized as a quarter start date

This allows Airbnb and Bond data to be matched by:
- `area_code`
- `TimeFrame`

```python
christ_area["Pub_date"] = pd.to_datetime(
    christ_area["Pub_date"]
)

christ_area["TimeFrame"] = (
    christ_area["Pub_date"]
    .dt.to_period("Q")
    .dt.start_time
)
```

```python
christ_area.info()
```

```python
print("Dwelling Types:")
print(df_tnc["Dwelling Type"].unique())

print("\nNumber of Beds:")
print(df_tnc["Number Of Beds"].unique())
```

```python
print(
    df_tnc[
        (df_tnc["Dwelling Type"] == "ALL") &
        (df_tnc["Number Of Beds"] == "ALL")
    ].shape
)
```

##  Filtering to the relevant Bond records
It keeps only the Bond rows where:
- `Dwelling Type == "ALL"`
- `Number Of Beds == "ALL"`

This is useful because the comparison is done at the broadest aggregate level for consistent time and area comparisons.

```python
df_tnc_all = df_tnc[
    (df_tnc["Dwelling Type"] == "ALL") &
    (df_tnc["Number Of Beds"] == "ALL")
].copy()
```

```python
print("Rows in original Bond data:", len(df_tnc))
print("Rows in ALL/ALL Bond data:", len(df_tnc_all))
```

```python
duplicate_keys = df_tnc_all.duplicated(
    ["area_code", "TimeFrame"]
).sum()

print("Duplicate area_code + TimeFrame:", duplicate_keys)
```

##  Merging datasets
The notebook merges Airbnb data with Bond data using:
- `area_code`
- `TimeFrame`

This creates a single table with Airbnb listing information plus Bond median rent and related market statistics.


```python
merged_data = christ_area.merge(
    df_tnc_all,
    on=["area_code", "TimeFrame"],
    how="left",
    validate="many_to_one"
)
```

```python
df_tnc_all.info()
```

## Validation and quality checks
Before exporting, the notebook verifies:
- whether the merge produced matches
- whether there are duplicate `area_code + TimeFrame` combinations
- whether the merged data contains missing values

```python
print("Before merge:", len(christ_clean))
print("After merge:", len(merged_data))
```

```python
print(
    "Airbnb rows with Bond data:",
    merged_data["Median Rent"].notna().sum()
)

print(
    "Airbnb rows without Bond data:",
    merged_data["Median Rent"].isna().sum()
)
```

```python
print(
    df_tnc_all.duplicated(
        ["area_code", "TimeFrame"]
    ).sum()
)
```

##  Saving outputs
The final merged dataset is exported as CSV files 

```python
merged_data.to_csv(r"../Output/christ_bond_merged.csv", index=False)
```
