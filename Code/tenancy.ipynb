##I imported Pandas and Path to support data analysis and file path management.
These libraries help load, process, and manage the datasets efficiently.


```python
#import libraries
import pandas as pd
from pathlib import Path
```

##I loaded the Tenancy Services dataset and checked its shape and first few rows.
This confirms that the file was loaded correctly and provides an initial view of the data.


```python
#Load dataset
df = pd.read_csv(r"C:\Users\patel\OneDrive\Desktop\New folder\Detailed-Quarterly-Tenancy-Q1-2020-Q3-2026.csv")

print(df.shape)
df.head()
```

##I converted the `TimeFrame` column to datetime format and checked for invalid or missing dates.
This ensures the time values are valid before performing time-based analysis.


```python
# Convert TimeFrame to datetime
df["TimeFrame"] = pd.to_datetime(df["TimeFrame"], errors="coerce")

# Check for invalid dates
if df["TimeFrame"].isna().any():
    raise ValueError("Invalid or missing TimeFrame values found.")

print("TimeFrame converted successfully.")
```

##I filtered the dataset to keep records from October 2025 to June 2026 and checked the number of rows before and after filtering.
This ensures the analysis uses only the required timeframe and confirms how many records were removed.


```python
#  Filter timeframe Keep only Oct 2025 to Jun 2026
bond = df[
    (df["TimeFrame"] >= "2025-10-01") &
    (df["TimeFrame"] < "2026-07-01")
].copy()

print("Rows before:", len(df))
print("Rows after:", len(bond))
print("Rows removed:", len(df) - len(bond))

bond["TimeFrame"].value_counts().sort_index()
```

##I checked each column for missing values after filtering the required timeframe.
This helps identify incomplete data before continuing with the analysis.


```python
#check missing values in each column
print("Missing values in each column:")
bond.isna().sum()
```

##I checked for missing locations, missing bed and rent values, and invalid location codes across each timeframe.
This helps identify data quality issues before using the dataset for analysis.


```python
# Data Quality investigate
checks = pd.DataFrame({
    "TimeFrame": bond["TimeFrame"],
    "Missing Location": bond["Location Id"].isna(),
    "Missing Beds": bond["Number Of Beds"].isna(),
    "Missing Rent": bond["Median Rent"].isna(),
    "Location -99": bond["Location Id"] == -99
})

checks.groupby("TimeFrame").sum()
```

##I checked the numerical variables for negative bond values, non-positive rents, incorrect quartile ordering, and duplicate rows.
These checks help confirm that the numerical data is valid and consistent before analysis.


```python
# validate numerical data
bond_columns = [
    "Total Bonds",
    "Active Bonds",
    "Closed Bonds"
]

rent_columns = [
    "Median Rent",
    "Geometric Mean Rent",
    "Lower Quartile Rent",
    "Upper Quartile Rent"
]

print("Negative bond values:")
print((bond[bond_columns] < 0).sum())

print("\nNon-positive rent values:")
print((bond[rent_columns] <= 0).sum())

invalid_quartiles = (
    (bond["Lower Quartile Rent"] > bond["Median Rent"]) |
    (bond["Median Rent"] > bond["Upper Quartile Rent"])
).sum()

print("\nInvalid quartile ordering:", invalid_quartiles)

print("Duplicate rows:", bond.duplicated().sum())
```

##I removed records with missing or invalid `Location Id` values to improve data quality.
I then checked the remaining missing values to confirm the cleaned dataset was suitable for analysis.


```python
# clean data set 
# Remove missing Location Id
cleaned = bond.dropna(subset=["Location Id"]).copy()

# Remove special Location Id = -99
cleaned = cleaned[cleaned["Location Id"] != -99].copy()

print("Final cleaned shape:", cleaned.shape)

# Check remaining missing values
cleaned.isna().sum()
```

##I checked for duplicate combinations of the key columns to ensure each record is uniquely identified.
This helps prevent duplicate records from affecting the accuracy of the analysis.


```python
# verify unique keys
key_columns = [
    "TimeFrame",
    "Location Id",
    "Dwelling Type",
    "Number Of Beds"
]

duplicates = cleaned.duplicated(subset=key_columns).sum()

print("Duplicate key combinations:", duplicates)
```

##I saved the cleaned dataset as a CSV file in the processed data folder.
This preserves the cleaned data for reproducible analysis and future use.


```python
#save cleaned dataset
output = Path("data/processed/rental_bond_cleaned_2025_10_to_2026_04.csv")

output.parent.mkdir(parents=True, exist_ok=True)

cleaned.to_csv(
    output,
    index=False,
    date_format="%Y-%m-%d"
)

print("Dataset saved successfully!")
print("Location:", output)
```

##I reloaded the saved dataset and compared its shape and column names with the cleaned dataset.
This confirms that the exported file was saved correctly without losing columns or rows.



```python
# verify export
saved = pd.read_csv(output)

print("Saved dataset shape:", saved.shape)

assert saved.shape == cleaned.shape
assert saved.columns.tolist() == cleaned.columns.tolist()

saved.head()
```
