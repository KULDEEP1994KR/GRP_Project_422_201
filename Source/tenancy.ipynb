```python
#import libraries
import pandas as pd
from pathlib import Path
```

```python
#Load dataset
df = pd.read_csv(r"C:\Users\patel\OneDrive\Desktop\new repo\GRP_Project_422_201\Detailed-Quarterly-Tenancy-Q1-2020-Q3-2026.csv")

print(df.shape)
df.head()
```

```python
# Convert TimeFrame to datetime
df["TimeFrame"] = pd.to_datetime(df["TimeFrame"], errors="coerce")

# Check for invalid dates
if df["TimeFrame"].isna().any():
    raise ValueError("Invalid or missing TimeFrame values found.")

print("TimeFrame converted successfully.")
```

```python
#  Filter timeframe Keep only Oct 2025 to Jun 2026
bond = df[
    (df["TimeFrame"] >= "2025-10-01") &
    (df["TimeFrame"] < "2026-07-01")
].copy()

print("Rows before:", len(df))
print("Rows after:", len(bond))
print("Rows removed:", len(df) - len(bond))

bond["TimeFrame"].value_counts().sort_index()
```

```python
#check missing values in each column
print("Missing values in each column:")
bond.isna().sum()
```

```python
# Data Quality investigate
checks = pd.DataFrame({
    "TimeFrame": bond["TimeFrame"],
    "Missing Location": bond["Location Id"].isna(),
    "Missing Beds": bond["Number Of Beds"].isna(),
    "Missing Rent": bond["Median Rent"].isna(),
    "Location -99": bond["Location Id"] == -99
})

checks.groupby("TimeFrame").sum()
```

```python
# validate numerical data
bond_columns = [
    "Total Bonds",
    "Active Bonds",
    "Closed Bonds"
]

rent_columns = [
    "Median Rent",
    "Geometric Mean Rent",
    "Lower Quartile Rent",
    "Upper Quartile Rent"
]

print("Negative bond values:")
print((bond[bond_columns] < 0).sum())

print("\nNon-positive rent values:")
print((bond[rent_columns] <= 0).sum())

invalid_quartiles = (
    (bond["Lower Quartile Rent"] > bond["Median Rent"]) |
    (bond["Median Rent"] > bond["Upper Quartile Rent"])
).sum()

print("\nInvalid quartile ordering:", invalid_quartiles)

print("Duplicate rows:", bond.duplicated().sum())
```

```python
# clean data set 
# Remove missing Location Id
cleaned = bond.dropna(subset=["Location Id"]).copy()

# Remove special Location Id = -99
cleaned = cleaned[cleaned["Location Id"] != -99].copy()

print("Final cleaned shape:", cleaned.shape)

# Check remaining missing values
cleaned.isna().sum()
```

```python
# verify unique keys
key_columns = [
    "TimeFrame",
    "Location Id",
    "Dwelling Type",
    "Number Of Beds"
]

duplicates = cleaned.duplicated(subset=key_columns).sum()

print("Duplicate key combinations:", duplicates)
```

```python
#save cleaned dataset
output = Path("data/processed/rental_bond_cleaned_2025_10_to_2026_04.csv")

output.parent.mkdir(parents=True, exist_ok=True)

cleaned.to_csv(
    output,
    index=False,
    date_format="%Y-%m-%d"
)

print("Dataset saved successfully!")
print("Location:", output)
```

```python
# verify export
saved = pd.read_csv(output)

print("Saved dataset shape:", saved.shape)

assert saved.shape == cleaned.shape
assert saved.columns.tolist() == cleaned.columns.tolist()

saved.head()
```
