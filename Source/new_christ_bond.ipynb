```python
import requests
import pandas as pd
import time

API_KEY = "819797b85c5949fdb5a1dbc4cd468409"
LAYER_ID = 123515
URL = "https://koordinates.com/services/query/v1/vector.json"

# Get unique coordinates
coordinates = (
    christ_clean[["latitude", "longitude"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

print("Unique coordinates:", len(coordinates))

area_codes = []

start_time = time.time()

for i, row in coordinates.iterrows():

    latitude = row["latitude"]
    longitude = row["longitude"]

    params = {
        "key": API_KEY,
        "layer": LAYER_ID,
        "x": longitude,
        "y": latitude,
        "max_results": 1,
        "radius": 0,
        "with_field_names": "true"
    }

    try:
        response = requests.get(
            URL,
            params=params,
            timeout=30
        )

        if response.status_code == 200:

            data = response.json()

            features = (
                data["vectorQuery"]
                ["layers"][str(LAYER_ID)]
                ["features"]
            )

            if features:
                area_code = features[0]["properties"].get(
                    "SA22026_V1_00"
                )
            else:
                area_code = None

        else:
            print("Error:", response.status_code)
            area_code = None

    except Exception as e:
        print("Error:", e)
        area_code = None

    area_codes.append(area_code)

    # Show progress every 100 requests
    if (i + 1) % 100 == 0:
        elapsed = time.time() - start_time
        print(
            f"{i + 1} / {len(coordinates)} completed "
            f"({elapsed / 60:.1f} minutes)"
        )

print("Finished!")

coordinates["area_code"] = area_codes
```
