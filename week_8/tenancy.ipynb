```python
import pandas as pd

# Load the Tenancy Services dataset
tenancy = pd.read_csv(
    "Detailed-Quarterly-Tenancy-Q1-2020-Q3-2026.csv"
)

# Display the original dataset size
print("Original rows:", len(tenancy))
print("Original columns:", len(tenancy.columns))

# Display the column names
print("\nOriginal columns:")
print(tenancy.columns.tolist())

# Remove columns that are completely empty
tenancy_cleaned = tenancy.dropna(axis=1, how="all")

# Keep the required columns, including TimeFrame and Location Id
columns_to_keep = [
    "TimeFrame",
    "Location Id",
    "Dwelling Type",
    "Number Of Beds",
    "Total Bonds",
    "Active Bonds",
    "Closed Bonds",
    "Median Rent",
    "Geometric Mean Rent",
    "Upper Quartile Rent",
    "Lower Quartile Rent",
    "Log Std Dev Weekly Rent"
]

tenancy_cleaned = tenancy_cleaned[columns_to_keep]

# Remove completely duplicated rows only
tenancy_cleaned = tenancy_cleaned.drop_duplicates()

# Display missing values after cleaning
print("\nMissing values after cleaning:")
print(tenancy_cleaned.isna().sum())

# Display the cleaned dataset size
print("\nCleaned rows:", len(tenancy_cleaned))
print("Cleaned columns:", len(tenancy_cleaned.columns))

# Calculate the number of rows removed
rows_lost = len(tenancy) - len(tenancy_cleaned)

print("Rows lost:", rows_lost)

# Save the cleaned dataset
tenancy_cleaned.to_csv(
    "tenancy_cleaned.csv",
    index=False
)

print("\nCleaned tenancy dataset saved successfully.")
```
